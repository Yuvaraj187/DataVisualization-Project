```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

data = df.head()

plt.plot(data["release_year"], range(1, len(data) + 1))

plt.title("Release Year vs Number of Records")
plt.xlabel("Release Year")
plt.ylabel("Record Number")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

data = df.head()

plt.bar(data["release_year"], range(1, len(data) + 1))

plt.title("Release Year vs Number of Records")
plt.xlabel("Release Year")
plt.ylabel("Record Number")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

data = df.head()

type_count = data["type"].value_counts()

plt.pie(
    type_count,
    labels=type_count.index,
    autopct='%1.1f%%'
)

plt.title("Movie and TV Show Distribution")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

data = df.head()

plt.boxplot(data["release_year"])

plt.title("Release Year Distribution")
plt.ylabel("Release Year")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

plt.hist(df["release_year"], bins=20)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_movies&shows.csv")

plt.scatter(df["release_year"], range(len(df)))

plt.title("Release Year vs Number of Titles")
plt.xlabel("Release Year")
plt.ylabel("Title Index")

plt.show()
```
