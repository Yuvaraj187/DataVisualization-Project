```python

import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("fifa_players.csv")


print(df)
```

```python

players = df["name"]
ratings = df["overall_rating"]
```

```python

top_players = df.sort_values("overall_rating", ascending=False).head(10)

plt.plot(top_players["name"], top_players["overall_rating"], marker="o")
plt.title("Top 10 FIFA Players by Overall Rating")
plt.xlabel("Players")
plt.ylabel("Overall Rating")
plt.xticks(rotation=45)
plt.show()
```

```python

top_players = df.sort_values("overall_rating", ascending=False).head(10)

plt.bar(top_players["name"], top_players["overall_rating"])
plt.title("Top 10 FIFA Players by Overall Rating")
plt.xlabel("Players")
plt.ylabel("Overall Rating")
plt.xticks(rotation=45)
plt.show()
```

```python

foot_counts = df["preferred_foot"].value_counts()

plt.pie(
    foot_counts,
    labels=foot_counts.index,
    autopct="%1.1f%%"
)
plt.title("Preferred Foot Distribution")
plt.show()
```

```python

plt.hist(df["overall_rating"], bins=10)

plt.title("Distribution of Overall Ratings")
plt.xlabel("Overall Rating")
plt.ylabel("Frequency")
plt.show()
```

```python

plt.scatter(df["age"], df["overall_rating"])

plt.title("Age vs Overall Rating")
plt.xlabel("Age")
plt.ylabel("Overall Rating")
plt.show()
```

```python

top_players = df.sort_values("overall_rating", ascending=False).head(10)

plt.plot(
    top_players["name"],
    top_players["overall_rating"],
    label="Overall Rating",
    marker="o"
)

plt.plot(
    top_players["name"],
    top_players["potential"],
    label="Potential",
    marker="o"
)

plt.title("Overall Rating and Potential of Top 10 Players")
plt.xlabel("Players")
plt.ylabel("Rating")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
