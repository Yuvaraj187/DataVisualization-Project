```python
import pandas as pd
df = pd.read_csv("fifa_players.csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_value = df[df["value_euro"] > 500000]

print(high_value)
```

```python
left_foot = df[df["preferred_foot"] == "Left"]

print(left_foot)
```

```python
sorted_df = df.sort_values(
    by="overall_rating",
    ascending=False
)

print(sorted_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.hist(df["overall_rating"], bins=20)

plt.title("Distribution of Overall Ratings", fontsize=16)
plt.xlabel("Overall Rating")
plt.ylabel("Number of Players")

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
print("Mean =", df["overall_rating"].mean())
print("Median =", df["overall_rating"].median())
print("Mode =")
print(df["overall_rating"].mode())
```
