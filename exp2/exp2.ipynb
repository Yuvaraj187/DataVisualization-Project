```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("fifa_players.db")

query = "SELECT * FROM fifa_players"

df = pd.read_sql(query, conn)

print(df)

conn.close()
```

```python
df.to_csv("fifa_players_export.csv", index=False)

df.to_excel("fifa_players_export.xlsx", index=False)

df.to_json("fifa_players_export.json", orient="records")

import sqlite3

conn = sqlite3.connect("fifa_players.db")
df.to_sql("fifa_players", conn, if_exists="replace", index=False)
conn.close()
```
