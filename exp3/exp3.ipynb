```python
import pandas as pd

df = pd.read_csv("fifa_players.csv")

print(df.iloc[:,:7].isnull())
```

```python
print(df.iloc[:,:7].isnull().sum())
```

```python
df["value_euro"] = df["value_euro"].fillna(df["value_euro"].mean())
df["wage_euro"] = df["wage_euro"].fillna(df["wage_euro"].mean())
df["release_clause_euro"] = df["release_clause_euro"].fillna(df["release_clause_euro"].mean())

print(df)
```

```python
print(df.duplicated())

df.drop_duplicates(inplace=True)

print(df)
```

```python
Q1 = df["overall_rating"].quantile(0.25)
Q3 = df["overall_rating"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["overall_rating"] < lower) |
    (df["overall_rating"] > upper)
]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["overall_rating"]] = scaler.fit_transform(df[["overall_rating"]])

print(df[["overall_rating"]])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["preferred_foot"] = encoder.fit_transform(df["preferred_foot"])

print(df[["preferred_foot"]].head())
```
